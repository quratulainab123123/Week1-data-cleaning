```python
import pandas as pd
import io

# Check if the CSV file was uploaded
if 'uploaded' not in globals() or not uploaded:
    from google.colab import files
    uploaded = files.upload()

# Find the CSV file
filename = next(
    (name for name in uploaded if name.lower().endswith('.csv')),
    None
)

if filename is None:
    print("Please upload your applicants.csv file.")
else:
    # Load the dataset
    df = pd.read_csv(io.BytesIO(uploaded[filename]))

    # Show dataset information
    print("Dataset loaded successfully!")
    print("Number of rows and columns:", df.shape)

    print("\nFirst five rows:")
    display(df.head())

    print("\nColumn names:")
    print(df.columns.tolist())

    print("\nDataset information:")
    df.info()

    print("\nMissing values in each column:")
    print(df.isnull().sum())

    print("\nNumber of exact duplicate rows:")
    print(df.duplicated().sum())
```

```python
# Remove spaces from column names
df.columns = df.columns.str.strip()

# Remove rows with missing applicant names
if 'Applicant Name' in df.columns:
    df['Applicant Name'] = df['Applicant Name'].astype('string').str.strip()
    df['Applicant Name'] = df['Applicant Name'].replace('', pd.NA)
    df = df.dropna(subset=['Applicant Name'])

# Fill missing emails
if 'Email' in df.columns:
    df['Email'] = df['Email'].fillna('not_provided@example.com')

# Fill missing universities
if 'University' in df.columns:
    df['University'] = df['University'].fillna('Not Specified')

# Fill missing statuses
if 'Status' in df.columns:
    df['Status'] = df['Status'].fillna('Under Review')

print("Missing values handled.")
print(df.isnull().sum())
```

```python
# Remove rows with missing applicant names
df = df.dropna(subset=['Applicant Name'])

# Fill missing email addresses
df['Email'] = df['Email'].fillna('not_provided@example.com')

# Fill missing universities
df['University'] = df['University'].fillna('Not Specified')

# Fill missing statuses
df['Status'] = df['Status'].fillna('Under Review')

print("Missing values handled.")
print(df.isnull().sum())
```

```python
# Count exact duplicate rows
print("Exact duplicates:", df.duplicated().sum())

# Remove exact duplicate rows
df = df.drop_duplicates()

# Remove duplicate applications based on email
df = df.drop_duplicates(subset=['Email'], keep='first')

print("Duplicate rows removed.")
print("Remaining rows:", len(df))
```

```python
# Remove extra spaces
for col in ['Applicant Name', 'Email', 'Phone',
            'Domain Applied', 'University', 'Status']:
    df[col] = df[col].astype('string').str.strip()

# Standardize domain names
domain_mapping = {
    'web dev': 'Web Development',
    'webdev': 'Web Development',
    'web development': 'Web Development',
    'data science': 'Data Science',
    'data science ': 'Data Science',
    'ai': 'Artificial Intelligence',
    'artificial intelligence': 'Artificial Intelligence',
    'machine learning': 'Machine Learning',
    'ml': 'Machine Learning'
}

df['Domain Applied'] = (
    df['Domain Applied'].str.lower().map(domain_mapping)
    .fillna(df['Domain Applied'])
)

# Standardize status
df['Status'] = df['Status'].str.title()

print("Domain names:")
print(df['Domain Applied'].unique())

print("\nStatuses:")
print(df['Status'].unique())
```

```python
# Convert application dates
df['Application Date'] = pd.to_datetime(
    df['Application Date'],
    errors='coerce',
    dayfirst=True
)

# Remove dashes and spaces from phone numbers
df['Phone'] = (
    df['Phone']
    .str.replace('-', '', regex=False)
    .str.replace(' ', '', regex=False)
)

print("Data types:")
df.info()

print("\nMissing dates:")
print(df['Application Date'].isnull().sum())
Step 8: Write your data
```
